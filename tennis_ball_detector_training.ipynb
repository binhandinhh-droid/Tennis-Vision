```python
from roboflow import Roboflow
rf = Roboflow(api_key="VT4QKmGnmUFK8WcGIVhV")
project = rf.workspace("viren-dhanwani").project("tennis-ball-detection")
version = project.version(6)
dataset = version.download("yolov8")
print(f"Dataset đã tải về tại: {dataset.location}")
```

```python
# Kiểm tra và chỉnh file data.yaml
import yaml
import os

data_yaml_path = f"{dataset.location}/data.yaml"

with open(data_yaml_path, 'r') as f:
    data_config = yaml.safe_load(f)

print(data_config)

# Đảm bảo đường dẫn là tuyệt đối và đúng
data_config['train'] = f"{dataset.location}/train/images"
data_config['val'] = f"{dataset.location}/valid/images"
data_config['test'] = f"{dataset.location}/test/images"

with open(data_yaml_path, 'w') as f:
    yaml.dump(data_config, f)

print("Đã cập nhật data.yaml")
print(data_config)
```

```python
# Train model
from ultralytics import YOLO

model = YOLO("yolov8m.pt")

results = model.train(
    data=data_yaml_path,
    epochs=300,
    imgsz=1280,
    batch=32,
    patience=20,
    project="tennis_ball_detection",
    name="yolov8m_run",
    device=0
)
```

```python
# 5. Đánh giá model trên tập validation
metrics = model.val()
print(metrics)
```

```python
# 6. Dự đoán thử trên ảnh/video mới
model_path = "/content/runs/detect/tennis_ball_detection/yolov8n_run/weights/best.pt"
trained_model = YOLO(model_path)

results = trained_model.predict(
    source=f"{dataset.location}/test/images",  # hoặc đường dẫn ảnh/video của bạn
    conf=0.25,
    save=True
)
```
